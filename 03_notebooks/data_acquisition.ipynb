# Continental Skies — Delay Root Cause and Reliability Analysis

## Objective
This project analyzes U.S. airline flight-level operational data (BTS 
Reporting Carrier On-Time Performance, Jan 2024 – Dec 2025) to identify 
patterns in flight delays, cancellations, and diversions. The goal is to 
uncover root causes behind operational disruptions and assess airline/
airport reliability, in order to produce clear, evidence-based insights 
for operational monitoring and decision-making.

## Initial Steps
1. **Data Acquisition** — Downloaded 24 monthly raw data files (Jan 2024 
   – Dec 2025) from the BTS TranStats source, organized into `01_Raw_data/2024` 
   and `01_Raw_data/2025` folders.
2. **Extraction** — Extracted each monthly `.zip` file into its respective 
   year folder to obtain the raw CSV files.
3. **Data Inventory** — Listed all extracted CSV files, checked file sizes 
   and confirmed all 24 months of data are present before loading.
4. **Data Loading** — Reading and combining all 24 CSV files into a single 
   dataframe for further inspection and analysis.

```python
import numpy as np
import pandas as pd
from pathlib import Path
import zipfile
```

```python
PROJECT_PATH = Path.cwd().parent
print("Project_Path", PROJECT_PATH.resolve())
```

```python
RAW_PATH = PROJECT_PATH / "01_Raw_data"
RAW_PATH.mkdir(parents=True, exist_ok=True)
print("Raw_path", RAW_PATH.resolve())

DOCUMENTATION_PATH = PROJECT_PATH / "02_documentation"
DOCUMENTATION_PATH.mkdir(parents=True, exist_ok=True)
print("Documentation_path", DOCUMENTATION_PATH.resolve())

```

```python
RD_2024=RAW_PATH/"2024"
RD_2024.mkdir(parents=True, exist_ok=True)
print("2024", RD_2024.resolve())

RD_2025=RAW_PATH/"2025"
RD_2025.mkdir(parents=True, exist_ok=True)
print("2025", RD_2025.resolve())
```

```python
def list_zip_files(paths):
    zip_files = []
    for path in paths:
        zip_files.extend(list(path.glob("*.zip")))

    print(f"Total Zip Files: {len(zip_files)}")

    for file in zip_files:
        print("File Name: ", file.name)
        print("File Size: ", round(file.stat().st_size / (1024 * 1024), 2), "MB")
        print("File Location:", file.resolve())

    return zip_files
```

```python
zip_files = list_zip_files([RD_2024, RD_2025])
```

```python
for files in zip_files:
    with zipfile.ZipFile(files,"r") as z :
        z.extractall(files.parent)

    print(f'Extracted:{files.name} - {files.parent.resolve()}')
```

```python
csv_files = list(RAW_PATH.rglob("*.csv"))
for file in csv_files:
    print(file)
```

```python
combined_file = RAW_PATH / "csv_2024_2025.csv"
print("csv_2024_2025.csv", combined_file.resolve())

first_file = True

for file in csv_files:
    print("Processing:", file.name)

    for chunk in pd.read_csv(file, chunksize=100_000, low_memory=False):
        chunk.to_csv(
            combined_file,
            mode="w" if first_file else "a",
            index=False,
            header=first_file
        )
        first_file = False

print("Combined CSV created:", combined_file)
```
